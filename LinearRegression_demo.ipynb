## Import necessary Libries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

## Generate some dummy data

## $$ y = \beta_0 + \beta_1 * w + \varepsilon $$

```python
# Let' say X is "Hours Studied" and y is the " Exam Score"
np.random.seed(0)
x = 2.5 * np.random.randn(100, 1) + 1.5 # Array of 100 values
res = 0.5 * np.random.randn(100, 1) # Random noise
y = 2 + 0.3 * x + res # The actual relationship equation
```

```python
print('Hours Studied:\n', x)
```

```python
print('Exam Score :\n', y)
```

## Split the data into training and Testing Set
##### We train on 80% of the data and test on the remaining 20%

```python
x_train, x_test, y_train, y_test = train_test_split(x, y , test_size = 0.2, random_state = 42)
```

```python
x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

##  Create the Linear Regression Model

```python
model = LinearRegression()
```

```python
# Train the model
model.fit(x_train, y_train)
```

```python
# Make a predictions using the testing set
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
# Evaluate the result
print("Coefficients of the model :\n", model.coef_[0][0])
```

```python
print("Intercept:\n", model.intercept_[0])
```

```python
print("Mean squared error: %.2f\n" % mean_squared_error(y_test, y_pred))
```

```python
print("Coefficient of determination (R^2): %.2f\n" % r2_score(y_test, y_pred))
```

```python
# 7. Visualization 
plt.scatter(x_test, y_test, color="black", label="Actual Data")
plt.plot(x_test, y_pred, color="blue", linewidth=3, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Linear Regression Example")
plt.legend()
plt.show()
```
